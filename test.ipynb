```python
from quant import quantize_2bit, quantize_4bit
import torch
```

```python
def generate_random_weights(shape, dtype=torch.float32, low=-1.0, high=1.0, device='cpu'):
    return (high - low) * torch.rand(shape, dtype=dtype, device=device) + low
w = generate_random_weights((1, 10))
print(w)
```

```python
w_2bit = quantize_2bit(w)
w_4bit = quantize_4bit(w)
print(w_2bit)
print(w_4bit)
```
